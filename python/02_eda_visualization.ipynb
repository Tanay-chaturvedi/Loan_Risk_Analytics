# Exploratory Data Analysis (EDA)

The objective of Exploratory Data Analysis is to understand the distribution of loan approvals and investigate how applicant financial and demographic factors are associated with loan approval.

The analysis focuses on variables such as CIBIL score, annual income, loan amount, loan term, employment status, education, number of dependents, and asset values.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df= pd.read_csv(r"C:\Users\USER\OneDrive\Desktop\Loan-Credit-Analytics\dataset\loan_approval_dataset.csv")
```

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="loan_status"
)

plt.title("Loan Approval Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Applicants")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="loan_status",
    y="cibil_score"
)

plt.title("CIBIL Score vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("CIBIL Score")

plt.show()
```

```python
print(
    df.groupby("loan_status")["cibil_score"]
      .agg(["mean", "median", "min", "max"])
)
```

### CIBIL Score vs Loan Approval

CIBIL score shows a strong relationship with loan approval status. Approved applicants have a substantially higher average and median CIBIL score compared with rejected applicants.

The average CIBIL score for approved applicants is approximately 703, while rejected applicants have an average of approximately 429. The boxplot also shows a clear difference in the distribution of CIBIL scores between the two groups.

This indicates that credit score is likely to be an important feature for predicting loan approval.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="loan_status",
    y="income_annum"
)

plt.title("Annual Income vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Annual Income")

plt.show()
```

```python
print(
    df.groupby("loan_status")["income_annum"]
      .agg(["mean", "median", "min", "max"])
)
```

### Annual Income vs Loan Approval

Annual income shows very little difference between approved and rejected applicants. The mean annual income of approved applicants is approximately ₹50.26 lakh, while rejected applicants have a mean of approximately ₹51.14 lakh. The median values are also very similar.

This suggests that annual income alone does not strongly distinguish between approved and rejected loan applications in this dataset.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="loan_status",
    y="loan_amount"
)

plt.title("Loan Amount vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Loan Amount")

plt.show()
```

### Loan Amount vs Loan Approval

The distribution of loan amounts is relatively similar for approved and rejected applicants. Both groups have comparable median loan amounts and similar overall ranges.

This suggests that loan amount alone does not strongly distinguish between approved and rejected applications in this dataset.
